```python
# Student Marks, Percentage and Grade
name = input("Enter student name: ")

while True:
  n = int(input("Enter number of subjects: "))
  if n > 0:
    break
  print("Number of subjects must be at least 1.")

max_mark = float(input("Enter maximum marks per subject (e.g., 100): "))
total = 0

for i in range(1, n + 1):
  while True:
    mark = float(input(f"Enter mark for Subject {i}: "))
    if 0 <= mark <= max_mark:
      total += mark
      break
    print(f"Invalid mark. Please enter a value between 0 and {max_mark}.")

percentage = (total / (n * max_mark)) * 100

if percentage >= 90:
  grade = "A+"
  performance = "Outstanding"
elif percentage >= 80:
  grade = "A"
  performance = "Excellent"
elif percentage >= 70:
  grade = "B"
  performance = "Very Good"
elif percentage >= 60:
  grade = "C"
  performance = "Good"
elif percentage >= 50:
  grade = "D"
  performance = "Average"
else:
  grade = "F"
  performance = "Needs Improvement"

print("\n--- Academic Performance ---")
print("Student Name :", name)
print("Total Marks  :", total, "/", n * max_mark)
print("Percentage   :", round(percentage, 2), "%")
print("Grade        :", grade)
print("Performance  :", performance)
```
